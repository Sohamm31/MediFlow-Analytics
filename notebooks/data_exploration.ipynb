```python


def clean_hospital_data(file_name):
    df = pd.read_csv(DATA_DIR / file_name, sep=";")

    df.columns = df.columns.str.strip()
    df = df.dropna(axis=1, how="all")

    for col in df.columns:
        if "time" in col.lower() and (
            "arrival" in col.lower() or "exiting" in col.lower()
        ):
            df[col] = pd.to_datetime(
                df[col].astype(str).str.strip(),
                dayfirst=True,
                errors="coerce"
            )

    numeric_columns = [
        "Waiting Time (min)",
        "Waiting Time Triage (min)",
        "Utilization %",
        "Length of Stay (min)",
        "LOS without Queues (min)"
    ]

    for col in numeric_columns:
        if col in df.columns:
            df[col] = pd.to_numeric(
                df[col].astype(str).str.replace(",", ".", regex=False),
                errors="coerce"
            )

    return df
```

```python

stage_names = [
    "CheckPatientType",
    "Triage",
    "CheckBloodPressure",
    "Fill_Information",
    "MedicalRecord1",
    "MedicalRecord2",
    "MedicalRecord3",
    "MedicalRecord4",
    "OutPatientDepartment"
]

patient_flow = []

for stage in stage_names:
    df = cleaned_data[stage]

    patient_flow.append({
        "Stage": stage,
        "Records": len(df),
        "Unique Patients": df["Patient ID"].nunique()
    })

patient_flow_df = pd.DataFrame(patient_flow)
patient_flow_df
```

```python

stage_patient_sets = {
    stage: set(cleaned_data[stage]["Patient ID"])
    for stage in stage_names
}

for stage in stage_names:
    other_stages = set().union(*[
        ids for other_stage, ids in stage_patient_sets.items()
        if other_stage != stage
    ])

    overlap = stage_patient_sets[stage] & other_stages

    print(f"{stage}: {len(overlap)} patients also appear in other stages")
```

```python

sample_ids = cleaned_data["OutPatientDepartment"]["Patient ID"].head(5).tolist()

for patient_id in sample_ids:
    print(f"\n Patient {patient_id} ")

    for stage in stage_names:
        df = cleaned_data[stage]

        patient_record = df[df["Patient ID"] == patient_id]

        if not patient_record.empty:
            print(f"\n{stage}")
            print(patient_record.to_string(index=False))
```

```python

ticket = cleaned_data["CheckPatientType"]
outpatient = cleaned_data["OutPatientDepartment"]

journey = ticket[
    ["Patient ID", "Queue Ticket Machine Arrival Time"]
].merge(
    outpatient[["Patient ID", "Exiting Time", "Length of Stay (min)"]],
    on="Patient ID",
    how="inner"
)

journey["Calculated LOS (min)"] = (
    journey["Exiting Time"] -
    journey["Queue Ticket Machine Arrival Time"]
).dt.total_seconds() / 60

journey["LOS Difference (min)"] = (
    journey["Length of Stay (min)"] -
    journey["Calculated LOS (min)"]
).abs()

print("Patients matched:", len(journey))
print("\nLOS comparison:")
print(journey[[
    "Patient ID",
    "Length of Stay (min)",
    "Calculated LOS (min)",
    "LOS Difference (min)"
]].head(10).round(2))

print("\nAverage absolute difference:",
      journey["LOS Difference (min)"].mean().round(3))
```

```python

print(journey["LOS Difference (min)"].describe().round(3))

print("\nMost common differences:")
print(
    journey["LOS Difference (min)"]
    .round(2)
    .value_counts()
    .head(10)
)
```

```python

journey = journey.merge(
    outpatient[["Patient ID", "Patient Type"]],
    on="Patient ID",
    how="left"
)

print(
    journey.groupby("Patient Type")["LOS Difference (min)"]
    .agg(["count", "mean", "median", "min", "max"])
    .round(3)
)
```

```python

journey = journey.merge(
    outpatient[["Patient ID", "Satisfaction %", "New Patient?"]],
    on="Patient ID",
    how="left"
)

journey["Difference Group"] = journey["LOS Difference (min)"].round(1)

print("By new-patient status:")
print(
    journey.groupby("New Patient?")["LOS Difference (min)"]
    .agg(["count", "mean", "median"])
    .round(3)
)

print("\nBy satisfaction:")
print(
    journey.groupby("Satisfaction %")["LOS Difference (min)"]
    .agg(["count", "mean", "median"])
    .round(3)
)
```

```python

for stage in stage_names:
    df = cleaned_data[stage]

    arrival_cols = [col for col in df.columns if "arrival time" in col.lower()]
    exit_cols = [col for col in df.columns if "exiting time" in col.lower()]

    if not arrival_cols or not exit_cols:
        continue

    arrival_col = arrival_cols[0]
    exit_col = exit_cols[0]

    duration_col = next(
        (col for col in df.columns
         if "waiting time" in col.lower()),
        None
    )

    if duration_col is None:
        continue

    calculated = (
        df[exit_col] - df[arrival_col]
    ).dt.total_seconds() / 60

    difference = (df[duration_col] - calculated).abs()

    print(
        f"{stage}: "
        f"rows={len(df)}, "
        f"mean difference={difference.mean():.3f} min, "
        f"max difference={difference.max():.3f} min"
    )
```
